```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import pandasql as ps
```

```python
def run_sql(query):
    return ps.sqldf(query, globals())
```

```python
#Check data in SQL
query = """
SELECT *
FROM df
LIMIT 5
"""

run_sql(query)
```

```python
df = pd.read_csv('SampleSuperstore.csv')
print("DataFrame 'df' loaded successfully.")
```

```python
#Total Sales
query = """
SELECT SUM(Sales) AS Total_Sales
FROM df
"""

run_sql(query)
```

```python
#Sales by Region
query = """
SELECT Region,
       SUM(Sales) AS Total_Sales
FROM df
GROUP BY Region
ORDER BY Total_Sales DESC
"""

run_sql(query)
```

```python
#Sales by Category
query = """
SELECT Category,
       SUM(Sales) AS Total_Sales
FROM df
GROUP BY Category
ORDER BY Total_Sales DESC
"""

run_sql(query)
```

```python
#Top 10 Products by Sales
query = """
SELECT [Product Name],
       SUM(Sales) AS Total_Sales
FROM df
GROUP BY [Product Name]
ORDER BY Total_Sales DESC
LIMIT 10
"""

run_sql(query)
```

```python
#Sales by Customer Segment
query = """
SELECT Segment,
       SUM(Sales) AS Total_Sales
FROM df
GROUP BY Segment
ORDER BY Total_Sales DESC
"""

run_sql(query)
```

```python
#Top Cities by Sales
query = """
SELECT City,
       SUM(Sales) AS Total_Sales
FROM df
GROUP BY City
ORDER BY Total_Sales DESC
LIMIT 10
"""

run_sql(query)
```

```python
#Orders count by Ship Mode
query = """
SELECT [Ship Mode],
       COUNT(*) AS Total_Orders
FROM df
GROUP BY [Ship Mode]
ORDER BY Total_Orders DESC
"""

run_sql(query)
```

```python
#Sales by State
query = """
SELECT State,
       SUM(Sales) AS Total_Sales
FROM df
GROUP BY State
ORDER BY Total_Sales DESC
LIMIT 10
"""

run_sql(query)
```

```python
#Unique Customers
query = """
SELECT COUNT(DISTINCT [Customer ID]) AS Unique_Customers
FROM df
"""

run_sql(query)
```

```python
#Top Customers by Sales
query = """
SELECT [Customer Name],
       SUM(Sales) AS Total_Sales
FROM df
GROUP BY [Customer Name]
ORDER BY Total_Sales DESC
LIMIT 10
"""

run_sql(query)
```
